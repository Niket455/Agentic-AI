Ollama
It supports embedding models, making it possible to build retrieval augmented generation(RAG) applications that combine text prompts with existing documents or other data

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = OllamaEmbeddings(model="nomic-embed-text")
```

```python
r1 = embeddings.embed_documents(
    ["Alpha is the first letter of the Greek alphabet.", 
     "Beta is the second letter of the Greek alphabet."]
     )
```

```python
r1[1]
```

```python
embeddings.embed_query("what is the second letter of the Greek alphabet?")
```
