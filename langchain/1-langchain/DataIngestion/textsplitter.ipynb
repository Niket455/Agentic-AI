Recursive text splitter

It is a technique used in RAG to divide a large document into smaller chunks for creating embeddings and storing them in a vector database. It tries to split the text at natural boundaries in a specific order, usually **paragraph (`\n\n`) → line (`\n`) → space (` `) → character**. If a chunk is still larger than the specified `chunk_size`, it moves to the next smaller separator and splits again. For example, if `chunk_size=500`, it tries to keep each chunk around 500 characters without unnecessarily breaking sentences or words. **`chunk_overlap`** means some content from the end of one chunk is repeated at the beginning of the next chunk, which helps preserve context between chunks. It is commonly used in RAG because it creates meaningful chunks that can be retrieved more effectively when answering user questions.


```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=30) 
final_documents = text_splitter.split_documents(docs)
final_documents
```

```python
print(final_documents[0])
print(final_documents[1])
```

HTMLHeaderTextSplitter
It is a text splitter used to divide an HTML document based on its **HTML headings**, such as `<h1>`, `<h2>`, and `<h3>`. Instead of splitting the text only by character length, it keeps the **heading and its related content together**, preserving the document's structure and context. For example, if an HTML page has `<h1>Python</h1>`, `<h2>Functions</h2>`, and `<h2>Classes</h2>`, the splitter creates chunks according to these sections and stores the heading information as metadata. This is useful in RAG because when a relevant section is retrieved, the LLM also knows **which section of the document the information belongs to**.


```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string = """<!DOCTYPE html>
<html>
<head>
    <title>Data Engineering Guide</title>
</head>
<body>

    <h1>Data Engineering</h1>
    <p>
        Data engineering is the process of collecting, transforming,
        storing, and managing data so that it can be used for analysis.
    </p>

    <h2>ETL</h2>
    <p>
        ETL stands for Extract, Transform, and Load. Data is extracted
        from different sources, transformed into a useful format, and
        loaded into a target system such as a data warehouse.
    </p>

    <h2>Apache Spark</h2>
    <p>
        Apache Spark is a distributed processing framework used to process
        large amounts of data efficiently.
    </p>

    <h3>PySpark</h3>
    <p>
        PySpark is the Python API for Apache Spark. It allows developers
        to write Spark applications using Python.
    </p>

    <h1>Data Storage</h1>
    <p>
        Data can be stored in databases, data warehouses, data lakes,
        and cloud storage systems.
    </p>

    <h2>Data Warehouse</h2>
    <p>
        A data warehouse stores structured and processed data for
        reporting and analytics.
    </p>

    <h2>Data Lake</h2>
    <p>
        A data lake can store large amounts of structured, semi-structured,
        and unstructured data in its raw or processed form.
    </p>

</body>
</html>"""


header_to_split_on = [("h1","Header 1"), ("h2","Header 2"), ("h3","Header 3")]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=header_to_split_on)
html_header_splits = html_splitter.split_text(html_string)
html_header_splits
```

**Character Text Splitter** is a simple text-splitting technique that divides a large document into smaller chunks based mainly on a specified number of characters. For example, if we set `chunk_size=500`, the text is split into chunks of around 500 characters. Unlike `RecursiveCharacterTextSplitter`, it does not try multiple separators such as paragraphs, lines, and spaces to preserve the natural structure of the text. We can also use `chunk_overlap` to repeat some characters from the previous chunk in the next chunk, which helps maintain context. It is useful when we simply need to divide text into fixed-size pieces before creating embeddings for a RAG system.


**Recursive JSON Splitter** is a text splitter used to divide large **JSON data** into smaller chunks while trying to preserve the JSON's structure and hierarchy. Instead of simply splitting the JSON based on character count, it recursively goes through nested objects, arrays, and fields and creates smaller chunks when the data becomes too large. For example, a large JSON containing `customers`, `orders`, and `products` can be split into smaller sections while keeping related fields together. This is useful in **RAG** when JSON documents are too large to embed as a single document, because it helps create meaningful chunks while preserving the relationship between the data fields.

If the value is ot a nested json, but rather a very large string the string will not be split. If you need a large cap on the chunk size consider composing this with a Recursive Text Splitter on those chunks. There is an optional pre-processing step to split lists, by first converting themm to json(dict) and the splitting them as such.

How the text is split: json value.
How the chunk size is measured: by number of characters.

```python
import json
import requests

json_data = requests.get("https://jsonplaceholder.typicode.com/posts").json()
```

```python
json_data
```

```python
from langchain_text_splitters import RecursiveJsonSplitter

json_data = {"posts": json_data}

json_splitter = RecursiveJsonSplitter(max_chunk_size=200)
json_chunks = json_splitter.split_json(json_data)
```
